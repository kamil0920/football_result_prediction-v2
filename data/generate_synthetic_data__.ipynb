```python
df = pd.read_csv('../data/preprocessed/preprocessed_1.csv')
data = df.reset_index()
print(data.shape)
data
```

```python
from ydata.dataset import Dataset
from ydata.synthesizers.regular import RegularSynthesizer
from ydata_synthetic.synthesizers.base import ModelParameters, TrainParameters
from ydata.metadata import Metadata

# Wrap the DataFrame in a Dataset object
dataset = Dataset(data)

dataset_attrs = {
    "sortbykey": "date",
}

# Create metadata from the Dataset
metadata = Metadata(dataset, dataset_attrs=dataset_attrs)

# Specify numerical and categorical columns
num_cols = data.select_dtypes(include='number').columns.tolist()
cat_cols = data.select_dtypes(exclude='number').columns.tolist()

# Define model parameters
model_params = ModelParameters(batch_size=500, lr=2e-4, betas=(0.5, 0.9))

# Initialize the synthesizer
synthesizer = RegularSynthesizer()

# Train the synthesizer on the data
synthesizer.fit(X=dataset, metadata=metadata)

# Generate synthetic samples
synthetic_data = synthesizer.sample(1000)
synthetic_data.head()
```

```python
import ydata

if isinstance(synthetic_data, ydata.dataset.dataset.Dataset):
    synthetic_data = synthetic_data.to_pandas()

X = pd.concat([synthetic_data, data])
X = X.drop(["match_api_id", "index"], axis="columns")
y = [0] * len(synthetic_data) + [1] * len(data)
```

```python
X
```

```python
# Adversarial validation

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import roc_auc_score


model = RandomForestClassifier()

feature_cols = [col for col in X.columns if col not in
                    ["match_api_id", "result_match", "season", "stage", "date", "home_team", "away_team"]]

X = X[feature_cols]

cv_preds = cross_val_predict(model, X, y, cv=5, n_jobs=-1, method="predict_proba")
roc_adv_score = roc_auc_score(y_true=y, y_score=cv_preds[:, 1])
print(f"roc auc adv score: {roc_adv_score:0.3f}")
```
