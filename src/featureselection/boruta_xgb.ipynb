```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/preprocessed/imputed_data.csv')

feature_cols = [col for col in df.columns if col not in
                ["result_match", "stage", "season", "date", "home_team", "away_team", "match_api_id"]]

y = df[['result_match']]
X = df[feature_cols].copy()
```

```python
from xgboost import XGBClassifier
from boruta import BorutaPy

params = {'colsample_bytree': 0.8949861281692281,
          'gamma': 3.5920440761581354,
          'learning_rate': 0.2861004914177646,
          'max_depth': 12,
          'min_child_weight': 13.500372304245893,
          'n_estimators': 350,
          'reg_alpha': 0.1842741694829661,
          'reg_lambda': 0.008596007166608277,
          'subsample': 0.8552978617018039}

y_train_value_counts = y.value_counts()
ratio = y_train_value_counts[0] / y_train_value_counts[1]

model = XGBClassifier(
    **params,
    random_state=42,
    eval_metric="logloss",
    scale_pos_weight=ratio
)

boruta_selector = BorutaPy(estimator=model, n_estimators='auto', perc=50, max_iter=75, verbose=1)
boruta_selector.fit(X, y.squeeze())
selected_features = boruta_selector.support_
```

```python
Confirmed: 	18
Tentative: 	17
Rejected: 	10
```

```python
selected_data = X.columns[selected_features]
print(selected_data)
```
